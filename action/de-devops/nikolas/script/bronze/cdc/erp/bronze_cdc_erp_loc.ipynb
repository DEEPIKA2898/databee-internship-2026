# Bronze Layer: ERP Customer Location (Auto Loader)

**Pattern**: Auto Loader (Streaming Ingestion)  
**Trigger**: AvailableNow (Incremental Batch)

```python
from pyspark.sql.functions import current_timestamp

source_path = "/Volumes/workspace/bronze/raw_sources/source_erp/"
target_table = "workspace.bronze.erp_loc_a101_cdc"
checkpoint_path = "/Volumes/workspace/bronze/raw_sources/checkpoints/erp_loc"
schema_path = "/Volumes/workspace/bronze/raw_sources/checkpoints/erp_loc_schema"

query = (
    spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option("cloudFiles.inferColumnTypes", "true")
    .option("cloudFiles.schemaLocation", schema_path)
    .option("pathGlobFilter", "LOC_A101.csv")
    .load(source_path)
    .withColumn("ingestion_timestamp", current_timestamp())
    .drop("_rescued_data")
    .writeStream
    .format("delta")
    .outputMode("append")
    .option("checkpointLocation", checkpoint_path)
    .trigger(availableNow=True)
    .toTable(target_table)
)

query.awaitTermination()
print(f"✓ Ingestion complete. Total records: {spark.table(target_table).count():,}")
```
